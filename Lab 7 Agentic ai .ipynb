```python
import sqlite3

conn = sqlite3.connect("memory.db")

conn.execute("""
CREATE TABLE IF NOT EXISTS facts(
    key TEXT,
    value TEXT
)
""")

conn.commit()

print("Database created successfully.")
```

```python
def remember(key, value):
    conn.execute(
        "INSERT INTO facts VALUES (?, ?)",
        (key, value)
    )
    conn.commit()

remember("name", "Riya")
remember("favorite_subject", "Mathematics")

print("Saved!")
```

```python
def recall_all():
    rows = conn.execute(
        "SELECT key, value FROM facts"
    ).fetchall()

    return {k: v for k, v in rows}

memory = recall_all()

print("Recalled memory:", memory)

response = llm.invoke(
    f"""
You are a friendly assistant.

Memory:
Name: {memory['name']}
Favourite Subject: {memory['favorite_subject']}

Reply ONLY with a greeting in plain English.
Do NOT write code.
Do NOT use markdown.
Example:
Hello Riya! Great to see you again. Ready for some Mathematics today?
"""
)

print(response.content)
```
